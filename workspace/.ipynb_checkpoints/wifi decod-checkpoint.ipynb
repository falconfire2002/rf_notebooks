#wifi constelation

```python
import io
import threading
import time
import numpy as np
import matplotlib
matplotlib.use("Agg")  # Headless renderer, completely thread-safe
import matplotlib.pyplot as plt
import adi
from IPython.display import display
import ipywidgets as widgets

#wifi standards
#units
us=1e-6
KHz=1e3
MHz=1e6
GHz = URI = 1e9


fft_size = 256
subcarrier_space = 78.125 * KHz  # kHz
symbol_duration = 12.8 * us      # us
guard_int = [0.8 * us, 1.6 * us, 3.2 * us]  # us
activeSubcarriers = 242

samplerate = int(fft_size * subcarrier_space)  # 20 MSps
symbol_samples = int(round(symbol_duration * samplerate))
gi_samples = [int(round(gi * samplerate)) for gi in guard_int]

# 802.11ax 20 MHz Subcarrier Maps (centred at DC = 0, range -128 to +127)
pilot_indices = np.array([-116, -90, -48, -22, 22, 48, 90, 116])
active_indices = np.concatenate([np.arange(-122, -1), np.arange(2, 123)])  # 242 tones
data_indices = np.setdiff1d(active_indices, pilot_indices)

# Convert signed subcarrier indices (-128..+127) to standard NumPy FFT array indices (0..255)
data_fft_bins = (data_indices + fft_size) % fft_size
pilot_fft_bins = (pilot_indices + fft_size) % fft_size

channel_center_freq = [ f * MHz for f in [2412, 2417, 2422, 2427, 2432, 2437, 2442, 2447, 2452, 2457, 2462, 2467, 2472, 2484]]
channel_options = [(f"Channel {ch} ({freq / MHz:.0f} MHz)", freq)for ch, freq in enumerate(channel_center_freq, start=1)]
channel_dropdown = widgets.Dropdown(options=channel_options,value=channel_center_freq[5],description="2.4 GHz Ch:",disabled=False,)

# --- UI Widgets ---
slider_layout = widgets.Layout(width='600px')
channel_dropdown = widgets.Dropdown(options=channel_options, value=channel_center_freq[5], description="2.4 GHz Ch:", disabled=False)
gain_slider = widgets.IntSlider(value=40, min=0, max=73, step=1, description="RX Gain (dB):")
gi_dropdown = widgets.Dropdown(options=[("0.8 µs", 0), ("1.6 µs", 1), ("3.2 µs", 2)], value=0, description="Guard Int:")
start_button = widgets.ToggleButton(value=False, description="Start Stream", button_style="success", icon="play")
update_period = widgets.FloatSlider(value=0.5, min=0.1, max=2.0, step=0.1, description='Period (s):', style={'description_width': 'initial'}, layout=slider_layout)
status_label = widgets.Label(value="Status: Idle")
output_image = widgets.Image(format="png", width=800, height=400)
constellation_image = widgets.Image(format="png", width=400, height=400)

dashboard_layout = widgets.VBox([
    widgets.HBox([channel_dropdown, gain_slider, gi_dropdown, start_button]),
    status_label,
    update_period,
    widgets.HBox([output_image, constellation_image])
])

# --- Helper Functions ---
def extract_symbol_iq(rx_symbol_block, gi_len):
    """
    Removes the guard interval, performs the 256-point FFT,
    and returns the complex data and pilot subcarriers.
    """
    if len(rx_symbol_block) < (gi_len + symbol_samples):
        return None, None
    
    # 1. Strip the cyclic prefix (guard interval)
    useful_time_samples = rx_symbol_block[gi_len : gi_len + symbol_samples]
    
    # 2. Convert time-domain IQ to frequency-domain subcarriers
    freq_bins = np.fft.fft(useful_time_samples, n=fft_size)
    
    # 3. Extract data and pilot subcarriers
    data_iq = freq_bins[data_fft_bins]
    pilot_iq = freq_bins[pilot_fft_bins]
    
    return data_iq, pilot_iq

    
def render_plot_bytes(x_data, psd, center_freq, bw_mhz):
    fig, ax = plt.subplots(figsize=(10, 4), dpi=80)
    ax.plot(x_data, psd, color='cyan', linewidth=1)
    ax.set_facecolor('#111111')
    ax.set_title(f"PlutoSDR Live PSD @ {center_freq / MHz:.1f} MHz (BW: {bw_mhz:.1f} MHz)", color='white')
    ax.set_xlabel("Frequency (MHz)", color='white')
    ax.set_ylabel("Relative Power (dB)", color='white')
    ax.tick_params(colors='white')
    ax.grid(True, which='both', linestyle='--', alpha=0.4)
    ax.set_ylim(-20, 80)
    fig.patch.set_facecolor('#1c1c1c')
    plt.tight_layout()

    buf = io.BytesIO()
    fig.savefig(buf, format='png', facecolor=fig.get_facecolor(), edgecolor='none')
    plt.close(fig)
    buf.seek(0)
    return buf.getvalue()

def render_constellation_bytes(data_iq):
    fig, ax = plt.subplots(figsize=(4, 4), dpi=80)
    if data_iq is not None and len(data_iq) > 0:
        ax.scatter(data_iq.real, data_iq.imag, color='magenta', s=5, alpha=0.6)
    ax.set_facecolor('#111111')
    ax.set_title("OFDM Data Constellation", color='white')
    ax.axhline(0, color='gray', linestyle='--', linewidth=0.5)
    ax.axvline(0, color='gray', linestyle='--', linewidth=0.5)
    ax.tick_params(colors='white')
    ax.grid(True, linestyle='--', alpha=0.3)
    fig.patch.set_facecolor('#1c1c1c')
    plt.tight_layout()

    buf = io.BytesIO()
    fig.savefig(buf, format='png', facecolor=fig.get_facecolor(), edgecolor='none')
    plt.close(fig)
    buf.seek(0)
    return buf.getvalue()

# --- Streaming Thread Loop ---
stream_thread = None

def rx_stream_worker():
    global start_button
    try:
        sdr = adi.Pluto("ip:192.168.2.1")
        sdr.sample_rate = int(samplerate)
        #rx setup
        sdr.rx_lo = int(channel_dropdown.value)
        sdr.rx_rf_gain_control_mode = "manual"
        sdr.rx_hardwaregain_chan0 = int(gain_slider.value)
        #tx setup (for rx only apps this is set like this to make sure tx is muted ad936x chips are notorios for dc breakthrough and noise so 0 streams are effective to shut it up)
        sdr.tx_lo = int(channel_dropdown.value)
        sdr.tx_rf_gain_control_mode = "manual"
        sdr.tx_hardwaregain_chan0=0

        num_samples = 4096
        mute_samples = np.zeros(num_samples, dtype=np.complex64)
        sdr.tx(mute_samples)
       
        # Use a larger buffer to capture enough history for correlation search
        sdr.rx_buffer_size = 1024 * 32
        
        status_label.value = f"Status: Streaming & Syncing on {sdr.rx_lo / MHz} MHz..."
        
        # Periodicity of L-STF short training symbols in 20 MHz (16 samples at 20 MSps)
        stf_period = 16 
        
        while start_button.value:
            sdr.rx_lo = int(channel_dropdown.value)
            sdr.rx_hardwaregain_chan0 = int(gain_slider.value)
            
            samples = sdr.rx()
            
            if len(samples) > stf_period * 4:
                # 1. Compute Power Spectral Density (PSD) for the display
                fft_vals = np.fft.fftshift(np.fft.fft(samples[:4096]))
                psd = 20 * np.log10(np.abs(fft_vals) + 1e-6)
                freqs = (np.fft.fftshift(np.fft.fftfreq(4096, 1 / samplerate)) + sdr.rx_lo) / MHz
                output_image.value = render_plot_bytes(freqs, psd, sdr.rx_lo, samplerate / MHz)
                
                # 2. Sliding Delay-16 Cross-Correlation for Packet Detection (L-STF repetition search)
                # This metric spikes when repetitive preamble training symbols are present
                corr = np.abs(np.sum(samples[:-stf_period] * np.conj(samples[stf_period:])))
                energy = np.sum(np.abs(samples[stf_period:])**2) + 1e-6
                metric = corr / energy
                
                # Find peak synchronization index exceeding threshold
                window_size = stf_period * 4
                correlation_profile = np.convolve(np.abs(samples), np.ones(window_size), mode='valid')
                
                if len(correlation_profile) > 0:
                    sync_idx = np.argmax(correlation_profile)
                    
                    # 3. Extract symbol block aligned to the detected boundary
                    selected_gi_len = gi_samples[gi_dropdown.value]
                    total_block_len = selected_gi_len + symbol_samples
                    
                    if sync_idx + total_block_len <= len(samples):
                        aligned_block = samples[sync_idx : sync_idx + total_block_len]
                        data_iq, _ = extract_symbol_iq(aligned_block, selected_gi_len)
                        
                        # Simple pilot-based phase tracking to stabilize the constellation
                        if data_iq is not None and len(data_iq) > 0:
                            # Remove residual random phase offset via mean angle alignment
                            phase_offset = np.angle(np.mean(data_iq))
                            data_iq = data_iq * np.exp(-1j * phase_offset)
                            
                        constellation_image.value = render_constellation_bytes(data_iq)
                    
            time.sleep(update_period.value)
            
        sdr = None
        status_label.value = "Status: Stopped"
    except Exception as e:
        status_label.value = f"Status: Error - {str(e)}"
        start_button.value = False

def on_start_button_change(change):
    global stream_thread
    if change['new']:
        status_label.value = "Status: Initializing PlutoSDR..."
        stream_thread = threading.Thread(target=rx_stream_worker, daemon=True)
        stream_thread.start()
    else:
        status_label.value = "Status: Stopping..."

start_button.observe(on_start_button_change, names='value')

# --- Display Dashboard ---
display(dashboard_layout)
```
